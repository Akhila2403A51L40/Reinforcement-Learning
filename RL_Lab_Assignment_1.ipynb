```python
# Greedy Multi-Armed Bandit using Arithmetic Mean

num_machines = int(input("Enter number of machines: "))

machines = {}

for i in range(num_machines):
    name = input(f"\nEnter name of Machine {i + 1}: ")
    rewards = list(map(float, input("Enter observed rewards separated by spaces: ").split()))
    machines[name] = rewards

print("\n--- Estimated Q Values ---")

q_values = {}

for machine, rewards in machines.items():
    q = sum(rewards) / len(rewards)
    q_values[machine] = q

    print(f"Machine {machine}")
    print(f"Rewards: {[int(r) if r.is_integer() else r for r in rewards]}")
    print(f"Q({machine}) = {q:.2f}")
    print()

selected_machine = max(q_values, key=q_values.get)

print(f"Greedy Selected Machine: {selected_machine}")
```

```python
# Greedy Multi-Armed Bandit using Incremental Update

num_machines = int(input("Enter number of machines: "))

machines = {}

# Input machine names and rewards
for i in range(num_machines):
    name = input(f"\nEnter name of Machine {i + 1}: ")
    rewards = list(map(float, input("Enter observed rewards separated by spaces: ").split()))
    machines[name] = rewards

# Initialize Q and N
Q = {}
N = {}

for machine in machines:
    Q[machine] = 0.0
    N[machine] = 0

# Process each reward incrementally
for machine, rewards in machines.items():

    print(f"\nProcessing Machine {machine}")

    for reward in rewards:

        # Increment number of plays
        N[machine] += 1

        # Incremental update formula
        Q[machine] = Q[machine] + (reward - Q[machine]) / N[machine]

        print(
            f"Reward = {reward:.0f}, "
            f"N({machine}) = {N[machine]}, "
            f"Q({machine}) = {Q[machine]:.2f}"
        )

# Display final values
print("\n--- Final Estimated Q Values ---")

for machine in machines:
    print(f"Machine {machine}")
    print(f"Rewards: {machines[machine]}")
    print(f"Q({machine}) = {Q[machine]:.2f}")
    print(f"N({machine}) = {N[machine]}")
    print()

# Greedy selection
selected_machine = max(Q, key=Q.get)

print(f"Greedy Selected Machine: {selected_machine}")
```
