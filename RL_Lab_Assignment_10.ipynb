```python

# Assignment 10: Bellman Optimality and Value Iteration
# Deterministic Grid World

gamma = 0.9
theta = 0.001

states = ["S0", "S1", "G"]
actions = ["Up", "Down", "Left", "Right"]

# Transition mapping: (Next State, Reward)
transitions = {
    "S0": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S0", -1),
        "Right": ("S0", -1)
    },
    "S1": {
        "Up": ("S0", -1),
        "Down": ("S1", -1),
        "Left": ("S1", -1),
        "Right": ("G", 10)
    }
}

# 1. Display MDP information
print("STATES:", states)
print("ACTIONS:", actions)
print("DISCOUNT FACTOR:", gamma)
print("NORMAL REWARD: -1")
print("WALL REWARD: -1")
print("GOAL REWARD: +10")

# 2. Verify transitions
print("\nTRANSITION MAPPING")
for s in ["S0", "S1"]:
    for a in actions:
        ns, r = transitions[s][a]
        print(s, "--", a, "-->", ns, "Reward:", r)

# 3. Initialize state values
V = {"S0": 0.0, "S1": 0.0, "G": 0.0}
print("\nINITIAL STATE VALUES:", V)

# 4. Value Iteration using synchronous updates
iteration = 0

print("\nITERATION RESULTS")
print("Iteration\tV(S0)\t\tV(S1)\t\tDelta")

while True:
    iteration += 1
    new_V = {"G": 0.0}

    for s in ["S0", "S1"]:
        values = []

        for a in actions:
            ns, reward = transitions[s][a]
            q = reward + gamma * V[ns]
            values.append(q)

        new_V[s] = max(values)

    delta = max(
        abs(new_V[s] - V[s])
        for s in ["S0", "S1"]
    )

    print(
        iteration, "\t\t",
        round(new_V["S0"], 4), "\t\t",
        round(new_V["S1"], 4), "\t\t",
        round(delta, 4)
    )

    V = new_V

    if delta < theta:
        break

# 5. Display final state values
print("\nFINAL OPTIMAL STATE VALUES")
for s in states:
    print("V*(", s, ") =", round(V[s], 4))

# 6. Calculate action values
Q = {}

print("\nACTION VALUES Q*(s,a)")
for s in ["S0", "S1"]:
    Q[s] = {}

    for a in actions:
        ns, reward = transitions[s][a]
        Q[s][a] = reward + gamma * V[ns]
        print(
            "Q*(", s, ",", a, ") =",
            round(Q[s][a], 4)
        )

# 7. Extract optimal policy
print("\nOPTIMAL POLICY")
for s in ["S0", "S1"]:
    best_value = max(Q[s].values())

    best_actions = [
        a for a in actions
        if abs(Q[s][a] - best_value) < 1e-8
    ]

    print(s, ":", best_actions)

# 8. Verify Bellman Optimality
print("\nBELLMAN OPTIMALITY VERIFICATION")
for s in ["S0", "S1"]:
    max_q = max(Q[s].values())

    print(
        s,
        "V*(s) =", round(V[s], 4),
        "max Q*(s,a) =", round(max_q, 4),
        "Verified =", abs(V[s] - max_q) < 1e-8
    )

# 9. Result explanation
print("\nRESULT")
print("The agent starts from S0.")
print("The optimal action from S0 is Down.")
print("The optimal action from S1 is Right.")
print("The goal state G is terminal.")
print("Value Iteration stops when Delta is less than theta.")
```
