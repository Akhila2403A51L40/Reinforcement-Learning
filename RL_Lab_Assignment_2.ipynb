```python
import random

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

# Number of machines
M = 3

# Number of plays
N_play = 10

# Exploration probability
epsilon = 0.2

# Part B - Initialize N, R and Q tables
N = {machine: 0 for machine in machines}
R = {machine: [] for machine in machines}
Q = {machine: 0.0 for machine in machines}

# Store total reward
total_reward = 0

print("===== Epsilon-Greedy Multi-Armed Bandit =====")
print("Machines:", machines)
print("Number of plays:", N_play)
print("Epsilon:", epsilon)

# Part C - Epsilon-Greedy Decision
for play in range(1, N_play + 1):

    # Generate random number
    r = random.random()

    if r < epsilon:
        # Exploration
        selected_machine = random.choice(machines)
        decision = "Exploration"
    else:
        # Exploitation
        max_q = max(Q.values())

        # Select randomly among machines having the maximum Q-value
        best_machines = [
            machine for machine in machines
            if Q[machine] == max_q
        ]

        selected_machine = random.choice(best_machines)
        decision = "Exploitation"

    print("\n--------------------------------")
    print("Play:", play)
    print("Random number:", round(r, 4))
    print("Decision:", decision)
    print("Selected Machine:", selected_machine)

    # Part D - Reward Collection
    reward = float(input("Enter reward: "))

    # Part E - Q-Value Update
    N[selected_machine] += 1
    R[selected_machine].append(reward)

    # Incremental sample-average formula
    Q[selected_machine] = (
        Q[selected_machine]
        + (1 / N[selected_machine])
        * (reward - Q[selected_machine])
    )

    total_reward += reward

    print("Updated N:", N)
    print("Updated Q:", {m: round(Q[m], 2) for m in machines})

# Part F - Display Results
print("\n\n ======================================")
print("           FINAL RESULTS")

print("\nParticipation Count (N):")
for machine in machines:
    print(machine, ":", N[machine])

print("\nRewards (R):")
for machine in machines:
    print(machine, ":", R[machine])

print("\nFinal Q-Values:")
for machine in machines:
    print(machine, ":", round(Q[machine], 2))

print("\nTotal Reward after 10 plays:", total_reward)

# Analysis
best_machine = max(Q, key=Q.get)
print("\n\n ======================================")
print("             ANALYSIS")
print("Most selected machine:", max(N, key=N.get))
print("Best estimated machine:", best_machine)
print("Highest Q-value:", round(Q[best_machine], 2))
```
